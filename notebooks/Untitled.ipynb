# Update list of stocks

```python
import pandas as pd
from pathlib import Path
import configuration as cfg
```

```python
source_file = cfg.BASE_DIR / "my_data/list_of_stocks.csv"
```

# Stock list

```python
df = pd.read_csv(source_file)
```

```python
df.head()
```

# New format

```python
df_new = df.copy()
```

```python
df_new["ysymbol"] = df_new["symbol"]
```

```python
df_new["exchange"] = df_new.apply(lambda x: [], axis=1)
df_new["wkn"] = ""
df_new["notes"] = ""

df_new = df_new[["symbol","name","isin","wkn","ysymbol","gsymbol","exchange","notes"]].copy()
```

```python
df_new.head()
```

# Exporting

```python
df_new.to_csv(source_file)
```
